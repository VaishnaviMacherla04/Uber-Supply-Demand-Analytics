```python
import zipfile

# Unzip the file automatically using Python
with zipfile.ZipFile('archive (1).zip', 'r') as zip_ref:
    zip_ref.extractall('.')

print("Done! File is unzipped.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Read the unzipped dataset
df = pd.read_csv('ncr_ride_bookings.csv')

# 2. Print status breakdown
print("--- RIDE STATUS BREAKDOWN (%) ---")
status_col = [c for c in df.columns if 'Status' in c or 'status' in c][0]
print((df[status_col].value_counts(normalize=True) * 100).round(2))

# 3. Create and save the chart
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x=status_col, palette='magma')
plt.title('Ride Request Status Breakdown', fontsize=14, fontweight='bold')
plt.xlabel('Status')
plt.ylabel('Count')
plt.xticks(rotation=30)

plt.savefig('uber_booking_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```
